```python
import pandas as pd
import numpy as np
import plotly_express as px
```

```python
df= pd.read_csv('covid19.csv')
df
```

```python
df_melted = df.melt(id_vars=['Country/Region'], var_name='Date', value_name='Cases')
df_melted = df_melted.iloc[867:]
df_melted['Date'] = pd.to_datetime(df_melted['Date'])
```

```python
print(f"Jumlah negara yang terkena : {df['Country/Region'].nunique()}")
print(f"Jumlah Provinsi yang terkena : {df['Province/State'].nunique()}")
```

```python
df1 =df.iloc[:,4:].sum().reset_index(name='Jumlah')
df1 = df1.rename(columns={'index':'Date'})
df1['Date'] = pd.to_datetime(df1['Date'])
```

```python
df1_monthly = df1.resample('M', on='Date').sum().reset_index()
df1_yearly = df1.resample('Y', on='Date').sum().reset_index()
```

```python
df_daily = df.iloc[:,4:].sum().diff().reset_index(name='Jumlah')
df_daily = df_daily.rename(columns={'index':'Date'})
df_daily['Date'] = pd.to_datetime(df_daily['Date'])
```

```python
df_monthly = df_daily.resample('M', on='Date').sum().reset_index()
df_yearly = df_daily.resample('Y', on='Date').sum().reset_index()
```
